```python
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0.6)

prompt_template_name = PromptTemplate(
    input_variables =['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this."
)

name_chain = prompt_template_name | llm | StrOutputParser()

prompt_template_items = PromptTemplate(
    input_variables = ['restaurant_name'],
    template="""Suggest some menu items for {restaurant_name}"""
)

food_items_chain = prompt_template_items | llm | StrOutputParser()
```

#### Simple Sequential Chain

```python
from langchain_core.runnables import RunnableSequence

# Linking two chains: Output of name_chain becomes input of food_items_chain
chain = name_chain | food_items_chain

content = chain.invoke({"cuisine": "Indian"})
print(content)
```

```python
from langchain_core.runnables import RunnablePassthrough

chain = (
    {"cuisine": RunnablePassthrough()} 
    | RunnablePassthrough.assign(restaurant_name=name_chain) 
    | RunnablePassthrough.assign(menu_items=lambda x: food_items_chain.invoke(x["restaurant_name"]))
)
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from langchain_community.utilities import WikipediaAPIWrapper
import numexpr


# -------------------------
# Weather tool
# -------------------------
def get_weather(city: str) -> str:
    """Get the weather for a city."""
    return f"The weather in {city} is sunny."


# -------------------------
# Wikipedia tool
# -------------------------
wiki_api = WikipediaAPIWrapper(top_k_results=3)
def wikipedia_search(query: str) -> str:
    """Search Wikipedia for a query and return a summary."""
    return wiki_api.run(query)


# -------------------------
# Math tool
# -------------------------
def llm_math(question: str) -> str:
    """Safely evaluate a math expression in text using numexpr."""
    try:
        # extract digits/operators for evaluation
        # e.g., "what is 12 * 12?" → "12 * 12"
        # rudimentary cleanup
        expr = "".join(ch for ch in question if ch.isdigit() or ch in "+-*/(). ")
        result = numexpr.evaluate(expr).item()
        return f"{expr.strip()} = {result}"
    except Exception as e:
        return f"Could not evaluate math expression: {e}"


# -------------------------
# LLM setup
# -------------------------
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0,
)


# -------------------------
# Create agent (NEW TEMPLATE)
# -------------------------
agent = create_agent(
    model=llm,
    tools=[
        get_weather,
        wikipedia_search,
        llm_math,
    ],
    system_prompt="You are a helpful assistant that can answer factual questions, perform math, and check the weather."
)


# -------------------------
# Test
# -------------------------

agent.invoke({"messages":"What is the square root of 144?"})
agent.invoke({"messages":"Who is Nikola Tesla?"})
```

#### ConversationBufferWindowMemory

```python
from langchain.memory import ConversationBufferWindowMemory
# For Window memory in LCEL, you would use `RunnableWithMessageHistory` with a history class
# that trims messages, or manually trim the list in `get_session_history`. 
# For improved functionality, consider `TrimMessages` in modern LangChain 0.2+

# Here is how to use TrimMessages:
from langchain_core.messages import trim_messages
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

convo_history = ChatMessageHistory()
# (Assuming llm and convo_prompt are defined or we redefine them for safety)
# Redefine prompt for clarity since previous cells might have changed context
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
convo_prompt = ChatPromptTemplate.from_messages([
    ("system", "The following is a friendly conversation between a human and an AI."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}"),
])

trimmer = trim_messages(
    max_tokens=100,
    strategy="last",
    token_counter=llm,
    include_system=True,
)

chain_with_trimming = (
    RunnablePassthrough.assign(history=lambda x: trimmer.invoke(x["history"]))
    | convo_prompt
    | llm
    | StrOutputParser()
)

convo_window = RunnableWithMessageHistory(
    chain_with_trimming,
    lambda session_id: convo_history,
    input_messages_key="input",
    history_messages_key="history",
)

convo_window.invoke(
    {"input": "Who won the first cricket world cup?"},
    config={"configurable": {"session_id": "1"}}
)
```

```python
convo_window.invoke(
    {"input": "How much is 5+5?"},
    config={"configurable": {"session_id": "1"}}
)
```
