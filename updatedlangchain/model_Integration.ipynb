```python
# method 1
# here we are using init_chat_model function to initialize the chat model
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response
```

```python
# method 2
# here we are using ChatGroq class to initialize the chat model
import os
from langchain_groq import ChatGroq



model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response
```

```python
''' Streaming And Batch

 Streaming:
Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses.
Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time'''
```

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
# iterate over the model response in chunks. 
# streaming returns a generator object, then we will iterate over it and print the text.
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True) # end="|" is used to print the text in the same line with a "|" in between each chunk.
    # flush=True is used to print the text immediately without waiting for the buffer to fill up.
```

```python
model.invoke("Why do parrots have colorful feathers?")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)
```

```python
# Batch
'''Batching a collection of independent requests to a model can significantly improve performance and reduce costs, 
as the processing can be done in parallel:
'''
```

```python
# in batch we can send multiple prompts at once
# all of them will be processed in parallel
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response)
```

```python
model.batch(
    ["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={ # config is used to control the number of parallel calls at a time.
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
```
